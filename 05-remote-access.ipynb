# Remote System Access

During this course we have been working remotely on an SCRTP managed computer. Here, we have a short summary of some of the most useful remote tools.

## Tools

### `ssh`

As described in the introduction, this is the program that allows you to connect to remote machines. 
SSH is a protocol that uses public private keys. For an in-depth guide to `ssh` see the [linux handbook](https://linuxhandbook.com/ssh/).

`ssh-keygen` is the program that can create ssh key pairs for you. Use the man page for more information. We have described in the first part of this tutorial how to create a private public key pair. Alternatively, github provides a good overview of the process [here](https://docs.github.com/en/authentication/connecting-to-github-with-ssh/generating-a-new-ssh-key-and-adding-it-to-the-ssh-agent).

When connecting to a server, the ssh-agent checks your private key against the public key on the server, for additional security you can password protect your ssh key.


### `~/.ssh`

This is the default location for ssh files e.g. key pairs, configs and known hosts.

### `scp`

This is the Secured CoPy command. Use `man` for all the options. For an in-depth guide see the [linux handbook](https://linuxhandbook.com/scp-command/)

It is very similar to the `cp` command from earlier, however, it can access remote systems. 

First lets look at a 'remote file path':

In general a remote file path is this:

`<user>@<server>:<path>`

The latter half `<path>` is the same as `cp`.
However, you should consider where you are accessing, e.g. a `/` following the colon will start at the root of the directory, a `~` following it will start at `<user>`'s home.

`<user>` is the username for the remote server, you can leave it out if the remote username is the same as your username. 

`<server>` is the name we give to the remote computer. 
Sometimes this will be a `hostname` that on your local network will be `resolved`.
It can always be an ip address in fact when a hostname is `resolved` its resolved to an ip address.

To move a file from a to b then you use:

`scp <a: remote file path> <b: remote file path>`

Replacing these paths with remote file paths (or local paths) moves file a to b, it can also do recursive copying like we saw with `cp`.

For example, to transfer a tar.gz archive from Blythe to from wherever you are accessing Blythe, you would use:

`scp  <SCRTPUserName>@blythedata.scrtp.warwick.ac.uk:/springbrook/share/<group>/<YourName>/*tar.gz .`

### `rsync`

Remote Sync or Rsync is another way to copy from one location to another.
Rsync has one advantage, it only copies the files that have changed guaranteeing a shorter copy time.

Find more details about rsync in the [linux handbook](https://linuxhandbook.com/rsync-command-examples/)

### `wget`

`wget` is a fantastic tool for pulling files from the internet and extremely simple.

```bash
wget <URL>
```

All you need is the URL of the file you want to download and as always you can look at the man page to get more details.


That is it for the basic tools for working remotely.

### Setting up your Github SSH keys (locally and/or on godzilla)

Later in this tutorial (notebook 09), we will be giving a short introduction on how to use git and GitHub. If you are intending to use git more regularly or just to make git more convenient to use, you can create a private public key pair using the steps from the first notebook of this tutorial. Alternatively, github provides a good overview of the process [here](https://docs.github.com/en/authentication/connecting-to-github-with-ssh/generating-a-new-ssh-key-and-adding-it-to-the-ssh-agent).

Then, [add the public key to GitHub](https://docs.github.com/en/authentication/connecting-to-github-with-ssh/adding-a-new-ssh-key-to-your-github-account#adding-a-new-ssh-key-to-your-account) to set up your SSH access to github.

Once you are confident you have done this try:

```bash

ssh -T git@github.com

```

And you should see a message:

`Hi <USERNAME>! You've successfully authenticated, but GitHub does not provide shell access.`
